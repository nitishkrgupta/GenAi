```python
import os
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
prompt_template1 = ChatPromptTemplate.from_messages([
    ('human', 'You are a helpful assistant, write a story'),
    ('ai', 'output : {story}')
])
```

```python
llm_model1 = ChatGoogleGenerativeAI(
    model = 'gemini-3.1-flash-lite',
    api_key = os.environ['GEMINI_API_KEY'],
    temperature = 0.0
)
```

```python
parser = StrOutputParser()
```

```python
story_chain = prompt_template1 | llm_model1 | parser
```

```python
response = story_chain.invoke('write a story for 5 year old kid in 200 words')

print(response)
```

```python
prompt_template2 = ChatPromptTemplate.from_messages([
    ('human', 'You are a sentiment analysis assistant. Analyze the sentiment of the given text.'),
    ('ai', 'sentiment: {sentiment}')
])
```

```python
llm_model2 = ChatGoogleGenerativeAI(
    model = 'gemini-3.1-flash-lite',
    api_key = os.environ['GEMINI_API_KEY'],
    temperature = 0.0
)
```

```python
parser = StrOutputParser()
```

```python
sentiment_chain = prompt_template2 | llm_model2 | parser
```

```python
response = sentiment_chain.invoke('''I absolutely loved this restaurant! The food was delicious, the staff were extremely friendly, and the service was fast. The atmosphere was also wonderful. I would definitely visit again and recommend it to everyone.''')
```

```python
print(response)
```
