```python
prompt_template1 = ChatPromptTemplate.from_messages([
    ('system', 'You are a helpful assistant,'
      'translate the given text in English language'
      'return the translated text without adding extra words'
      'also detect the language'),
    ('human', '{text}')
])
```

```python
llm_model2 = ChatGoogleGenerativeAI(
    model = 'gemini-3.1-flash-lite',
    api_key = os.environ['GEMINI_API_KEY'],
    temperature = 0.0
)
```

```python
translate_chain = prompt_template1 | llm_model2 | parser
```

```python
prompt_template2 = ChatPromptTemplate.from_messages([
    ('system', 'You are a helpful assistant,'
      'write the summary of the given text'
      'also the detected the language'
      ),
    ('human', '{text}')
])
```

```python
llm_model3 = ChatGoogleGenerativeAI(
    model = 'gemini-3.1-flash-lite',
    api_key = os.environ['GEMINI_API_KEY'],
    temperature = 0.0
)
```

```python
summary_chain = prompt_template2 | llm_model3 | parser
```

```python
# summary = summary_chain.invoke({'text' : f'{translated_text}'})
# summary
```

```python
full_chain = translate_chain | summary_chain
```

```python
response = full_chain.invoke({'text' : "Machine Learning oru easyana subject, ithala niraya algorithms irruku ellane easy"})
response
```
