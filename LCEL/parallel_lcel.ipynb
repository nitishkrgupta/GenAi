```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
import os
from langchain_core.runnables import RunnablePassthrough, RunnableParallel
from langchain_community.document_loaders import PyPDFLoader, TextLoader
```

```python
parser = StrOutputParser()
```

```python
llm = ChatGoogleGenerativeAI(
    model = 'gemini-3.5-flash-lite',
    api_key = os.environ['GEMINI_API_KEY']
)
```

##### LCEL 1 - `Extracting the information from resume`

```python
skills_prompt = ChatPromptTemplate.from_messages([
    ('system', "extract the skills from the resume."),
    ('human', "resume : {resume}")
])
skill_chain = skills_prompt | llm | parser
```

##### LCEL - 2 `summarize about the project/ exp from resume`

```python
summarize_prompt = ChatPromptTemplate.from_messages([
    ('system', "summarize the experience from the resume"),
    ('human', "resume : {resume}")
])
summarize_chain = summarize_prompt | llm | parser
```

##### LCEL - 3 `Find the missing skills from resume for data science role`

```python
missing_skills_prompt = ChatPromptTemplate.from_messages([
    ('system', "for data science role find the missing skills from the resume"),
    ('human', "resume : {resume}")
])
missing_skills_chain = missing_skills_prompt | llm | parser
```

```python
final_chain = RunnableParallel(
    skills = skill_chain,
    summarize = summarize_chain,
    missing_skills = missing_skills_chain
)
```
