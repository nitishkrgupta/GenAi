```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
import os
```

# LCEL

#### Step 1 : Prompt Template

```python
prompt_template = ChatPromptTemplate.from_messages([
    ('human', "You're a helpful assistant, detect the Native communication Language from the given prompt" ),
    ('ai', "Output : {language}")
])
```

#### Step 2 : LLM

```python
llm_model = ChatGoogleGenerativeAI(
    model="gemini-3.1-flash-lite", # gemini-2.5-flash gemini-3.6-flash gemini-2.5-pro
    api_key = os.environ['GEMINI_API_KEY'],
    temperature=0.0
)
```

#### Step 3 : Parser output

```python
parser = StrOutputParser()
```

```python
language_chain = prompt_template | llm_model | parser
response = language_chain.invoke('''Machine learning is an easy subject,
its algorithms are also easy''')
# Machine Learning oru easyana subject, ithala niraya algorithms irruku ellane easy
```
