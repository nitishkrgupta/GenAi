```python
import os
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
user_input = '''
        AI systems can range from very simple rule-based programs to
        extremely complex neural networks with billions of parameters. A
        simple thermostat that adjusts temperature based on a rule is a
        primitive form of automation, but it is not considered true AI unless
        it can adapt and learn. Modern AI systems, on the other hand, can
        learn from data, adapt to new situations, and improve their
        performance over time without being explicitly reprogrammed.'''
```

```python
prompt_template = ChatPromptTemplate.from_messages([
    ('human', 'You are a helpful assistant, detect the language of the given input'),
    ('ai', 'Output : {Language}')
])
```

```python
llm_model1 = ChatGoogleGenerativeAI(
    model = 'gemini-3.1-flash-lite',
    api_key = os.environ['GEMINI_API_KEY'],
    temperature = 0.1
)
```

```python
parser = StrOutputParser()
```

```python
language_chain = prompt_template | llm_model1 | parser
```

```python
response = language_chain.invoke(user_input)
```

```python
model1_res = response
```

```python
prompt_template2 = ChatPromptTemplate.from_messages([
    ('human', "you are a multi language model translate the given input into English and don't add or repeat extra words just translate it"),
    ('ai', 'Output : {Translated_text}')
])
```

```python
llm_model2 = ChatGoogleGenerativeAI(
    model = 'gemini-3.5-flash',
    api_key = os.environ['GEMINI_API_KEY'],
    temperature = 0.0
)
```

```python
response = translate_chain.invoke(user_input)
model2_res = response
print(model2_res)
```
