#### Reading the data

```python
with open('D:\Study\pyspider\GenAI\Text_normalization\data.txt', 'r', encoding="utf-8") as file:
    data = file.read()
print(data)
```

#### convert to lowercase

```python
data = data.lower()
```

#### removing extra spaces

```python
words = data.split()
data = " ".join(words)
print(data)
```

#### expanding and abbreviation of words

```python
import contractions

data = contractions.fix(data)
data
```

#### Removing the punctuations 

```python
import string
import re

punc = string.punctuation
data = re.sub(f"[{punc}]", "", data)
```

```python
import emoji

data = emoji.replace_emoji(data, "")
data
```

#### removing special characters

```python
# for char in data:
#     if not char.isalnum() and not ord(char) == 32:
#         data = data.replace(char, "")
```

```python
data = re.sub("[^0-9a-zA-Z\s]", "", data)
```

#### correcting the words

```python
from textblob import TextBlob

corrected_words = TextBlob(data).correct()
# corrected_words = str(corrected_words)

corrected_words = TextBlob(data).correct().raw_sentences
data = " ".join(corrected_words)
print(data)
```

#### tokenization and lemmatization

```python
import spacy

nlp = spacy.load('en_core_web_sm')
tokens = nlp(data)

lemmatize_tokens = [token.lemma_ for token in tokens if not token.is_stop]
lemmatize_tokens
data = " ".join(lemmatize_tokens)
print(data)
```

#### chunking

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=100, 
    chunk_overlap=20, 
    separators=["\n\n","\n",".","!"," ",""])
text_splitter.split_text(data)
```
