```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.chat_history import InMemoryChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_core.output_parsers import StrOutputParser

llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite"
    )

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant."),
    MessagesPlaceholder(variable_name="history"),
    ("human", "{input}")
])

chain = prompt | llm | StrOutputParser()

store = {}

def get_session_history(session_id):
    if session_id not in store:
        store[session_id] = InMemoryChatMessageHistory() # python memory
    return store[session_id]


chain_with_memory = RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key="input",
    history_messages_key="history"
)

chat1 = chain_with_memory.invoke(
    {"input": "Hi!, My name is Nitish."},
    config={"configurable": {"session_id": "chat_1"}}
)
user1 = chain_with_memory.invoke(
    {"input" : "I am learning GenAi."},
    config={"configurable" : {"session_id" : "chat_1"}}
)
user1 = chain_with_memory.invoke(
    {"input": "Tell me about me"},
    config={"configurable": {"session_id": "chat_1"}}
)
chat2 = chain_with_memory.invoke(
    {"input" : "Hi"},
    config={"configurable" : {"session_id" : "chat2"}}
)
user2 = chain_with_memory.invoke(
    {"input" : "Tell me my name"},
    config={"configurable" : {"session_id" : "chat2"}}
)
user3 = chain_with_memory.invoke(
    {"input" : "Tell what i am learning"},
    config={"configurable" : {"session_id" : "chat2"}}
)
print(store)
print(chat1)
print(user1)
print()
print("---------SECOND CHAT RESULT------------")
print()
print(chat2)
print(user2)
print(user3)
```
